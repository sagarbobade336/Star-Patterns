#Number Pattern

```python
#number triangle
for i in range(5):
    # ptint space
    print(" "*(5-i),end='')
    for j in range(1,i+1):
        print(j,end=" ")
    print()    
```

```python
 #Reverse Pyramid
for i in range(5):
    print(" "*i,end="")
    for j in range(5-i,0,-1):
        print(j,end=" ")
    print()
```

```python
#Full Symmetrical Pyramid
rows = 5
for i in range(1, rows + 1):
    print("  " * (rows - i), end="")
    for j in range(1, i + 1):
        print(j, end=" ")
    for j in range(i - 1, 0, -1):
        print(j, end=" ")
    print()  # New lin
```

```python
#number square
for i in range(3):
    for j in range(1,4):
        print(j,end="")
    print()
```

```python
def tri(n):
    for i in range(n):
        for j in range(i+1):
            print(j,end="")
        print()    

tri(10)
```

```python
 #Floyd’s Triangle
num = 1  

for i in range(1, 5):  
    for j in range(i):
        print(num, end=' ')
        num += 1
    print()  
```

```python
def number_diamond(n):
    # Upper half 
    for i in range(1, n + 1):
        print(' ' * (n - i), end='')  # Spaces before numbers
        for j in range(1, i + 1):
            print(j, end=' ')
        print()

    # Lower half
    for i in range(n - 1, 0, -1):
        print(' ' * (n - i), end='')  # Spaces before numbers
        for j in range(1, i + 1):
            print(j, end=' ')
        print()

number_diamond(3)
```

```python
#zigzag triangle
def zigzag_triangle(n):
    num = 1
    for i in range(n):
        # Print leading spaces
        print(' ' * i, end='')

        # Print first number
        print(num, end='')

        num += 1

        # Calculate inner space (only if not last row)
        if i < n - 1:
            spaces = 2 * (n - i - 1) - 1
            print(' ' * spaces, end='')

            # Print second number
            print(num)
            num += 1
        else:
            print()  # Just move to the next line for last row


zigzag_triangle(4)
```
